```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
temperature = np.arange(0, 41, 1)
cooling = np.arange(0, 101, 1)  
```

```python
def trimf(x, abc): 
    a, b, c = abc  
    return np.maximum(np.minimum((x - a) / (b - a + 1e-9), (c - x) / (c - b + 1e-9)), 0)
```

```python
temp_cold = trimf(temperature, [0, 0, 20])
temp_warm = trimf(temperature, [10, 20, 30])
temp_hot  = trimf(temperature, [20, 40, 40])
```

```python
temp_cold = trimf(temperature, [0, 0, 20]) 
temp_warm = trimf(temperature, [10, 20, 30]) 
temp_hot  = trimf(temperature, [20, 40, 40])
```

```python
plt.figure(figsize=(8,4)) 
plt.plot(temperature, temp_cold, label='Cold') 
plt.plot(temperature, temp_warm, label='Warm') 
plt.plot(temperature, temp_hot, label='Hot') 
plt.title("Temperature Membership Functions") 
plt.xlabel("Temperature (°C)")
plt.ylabel("Membership Degree") 
plt.legend() 
plt.savefig("fuzzy1.png") 
plt.show()
```
